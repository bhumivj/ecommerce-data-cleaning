```python
from google.colab import files
import zipfile
import os

uploaded = files.upload()

with zipfile.ZipFile("archive.zip", "r") as zip_ref:
    zip_ref.extractall("shopping_data")

print("Files extracted:")
print(os.listdir("shopping_data"))
```

```python
import pandas as pd

df = pd.read_csv("shopping_data/Combined_dataset.csv")
print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
print("=== FIRST 5 ROWS ===")
print(df.head())

print("\n=== LAST 5 ROWS ===")
print(df.tail())

print("\n=== SHAPE (rows, columns) ===")
print(df.shape)

print("\n=== COLUMN NAMES ===")
print(df.columns.tolist())

print("\n=== DATA TYPES ===")
print(df.dtypes)
```

```python
print("=== MISSING VALUES BEFORE CLEANING ===")
print(df.isnull().sum())

# Drop rows where critical columns are missing
df = df.dropna(subset=['product_id', 'title', 'final_price'])

# Fill missing text columns
df['seller_name'] = df['seller_name'].fillna('Unknown')
df['category'] = df['category'].fillna('Unknown')

print("\n=== MISSING VALUES AFTER CLEANING ===")
print(df.isnull().sum())
```

```python
# Remove ₹ symbol, quotes, and commas so we can do math
df['final_price'] = df['final_price'].astype(str)
df['final_price'] = df['final_price'].str.replace('₹', '', regex=False)
df['final_price'] = df['final_price'].str.replace(',', '', regex=False)
df['final_price'] = df['final_price'].str.replace('"', '', regex=False)
df['final_price'] = pd.to_numeric(df['final_price'], errors='coerce')

# Drop rows where price couldn't be converted
df = df.dropna(subset=['final_price'])

print("final_price column cleaned. Sample values:")
print(df['final_price'].head())
```

```python
# Select only relevant columns
df = df[['product_id', 'title', 'rating', 'ratings_count',
         'initial_price', 'discount', 'final_price', 'category', 'seller_name']]

# Filter: only products with rating >= 4
df_filtered = df[df['rating'] >= 4].copy()

print("Rows with rating >= 4:", df_filtered.shape[0])
print(df_filtered.head())
```

```python
print("Rows before removing duplicates:", df_filtered.shape[0])

df_filtered = df_filtered.drop_duplicates()

print("Rows after removing duplicates:", df_filtered.shape[0])
```

```python
# total_amount = final_price × ratings_count
# ratings_count is used as quantity (number of times purchased/reviewed)
df_filtered['total_amount'] = df_filtered['final_price'] * df_filtered['ratings_count']

print("New column 'total_amount' added:")
print(df_filtered[['title', 'final_price', 'ratings_count', 'total_amount']].head())
```

```python
df_filtered.to_csv("cleaned_ecommerce_dataset.csv", index=False)
print("Saved successfully!")

# Download to your laptop
from google.colab import files
files.download("cleaned_ecommerce_dataset.csv")
```
